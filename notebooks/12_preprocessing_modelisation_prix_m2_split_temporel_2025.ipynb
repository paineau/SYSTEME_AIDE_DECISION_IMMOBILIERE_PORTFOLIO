```python
from pathlib import Path
import json
import warnings
import time

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.base import clone
from sklearn.cluster import KMeans
from sklearn.compose import ColumnTransformer
from sklearn.dummy import DummyRegressor
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor
from sklearn.feature_selection import mutual_info_regression
from sklearn.inspection import permutation_importance
from sklearn.linear_model import LinearRegression, Ridge, Lasso, HuberRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split, GridSearchCV, RandomizedSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

warnings.filterwarnings("ignore")
pd.set_option("display.max_columns", 250)
pd.set_option("display.max_rows", 120)

PROJECT_ROOT = Path.cwd().parents[0] if Path.cwd().name == "notebooks" else Path.cwd()

 
# ENTRÉES CANONIQUES — NOTEBOOK 11
 
# Le notebook 12 consomme directement les exports gouvernés du notebook 11.
# Les anciens fallbacks 00_07 / V3 / V8 / data/processed sont volontairement supprimés.

FEATURE_STORE_TAUX_DIR = PROJECT_ROOT / "outputs" / "feature_store_modelisation_taux"
FEATURE_STORE_TAUX_TABLES_DIR = FEATURE_STORE_TAUX_DIR / "tables"
FEATURE_STORE_TAUX_CONFIG_DIR = FEATURE_STORE_TAUX_DIR / "config"

INPUT_FEATURE_STORE_PATH = FEATURE_STORE_TAUX_TABLES_DIR / "dataset_modelisation_selected_taux.parquet"
INPUT_SCENARIOS_JSON_PATH = FEATURE_STORE_TAUX_CONFIG_DIR / "scenario_features_taux.json"
INPUT_FEATURE_BLOCKS_PATH = FEATURE_STORE_TAUX_CONFIG_DIR / "feature_blocks_selected_taux.json"
INPUT_MANIFEST_PATH = FEATURE_STORE_TAUX_CONFIG_DIR / "feature_store_manifest_taux.json"

 
# SORTIES — NOTEBOOK 12
 
OUTPUT_DIR = PROJECT_ROOT / "outputs" / "modelisation_prix_m2_taux_max_45000_split_temporel_2025"
FIG_DIR = OUTPUT_DIR / "figures"
TABLE_DIR = OUTPUT_DIR / "tables"
MODEL_DIR = OUTPUT_DIR / "models"

for p in [OUTPUT_DIR, FIG_DIR, TABLE_DIR, MODEL_DIR]:
    p.mkdir(parents=True, exist_ok=True)


def save_fig(name):
    plt.tight_layout()
    plt.savefig(FIG_DIR / name, bbox_inches="tight")
    plt.show()


required_inputs = {
    "dataset notebook 11": INPUT_FEATURE_STORE_PATH,
    "scénarios notebook 11": INPUT_SCENARIOS_JSON_PATH,
    "blocs sélectionnés notebook 11": INPUT_FEATURE_BLOCKS_PATH,
    "manifeste notebook 11": INPUT_MANIFEST_PATH,
}

for label, path in required_inputs.items():
    if not path.exists():
        raise FileNotFoundError(f"{label} introuvable : {path}")

with open(INPUT_MANIFEST_PATH, "r", encoding="utf-8") as f:
    feature_store_manifest = json.load(f)

source_file = INPUT_FEATURE_STORE_PATH
df_raw = pd.read_parquet(source_file).copy()

if "prix_m2" not in df_raw.columns and "prix_m2_model" in df_raw.columns:
    df_raw["prix_m2"] = df_raw["prix_m2_model"]
if "prix_m2" not in df_raw.columns and "log_prix_m2" in df_raw.columns:
    df_raw["prix_m2"] = np.exp(pd.to_numeric(df_raw["log_prix_m2"], errors="coerce"))

if "surface_reference" not in df_raw.columns and "surface_reference_model" in df_raw.columns:
    df_raw["surface_reference"] = df_raw["surface_reference_model"]

print("Projet                         :", PROJECT_ROOT)
print("Feature store notebook 11       :", FEATURE_STORE_TAUX_DIR)
print("Fichier chargé                  :", source_file)
print("Scénarios chargés depuis         :", INPUT_SCENARIOS_JSON_PATH)
print("Manifeste chargé depuis          :", INPUT_MANIFEST_PATH)
print("Version feature store            :", feature_store_manifest.get("version"))
print("Lignes                          :", f"{len(df_raw):,}")
print("Colonnes                        :", df_raw.shape[1])

display(df_raw.head(3))

df = df_raw.copy()

required_cols = ["prix_m2"]
missing_required = [c for c in required_cols if c not in df.columns]

if missing_required:
    raise ValueError(f"Colonnes obligatoires absentes : {missing_required}")

df["prix_m2"] = pd.to_numeric(df["prix_m2"], errors="coerce")

if "date_mutation" in df.columns:
    df["date_mutation"] = pd.to_datetime(df["date_mutation"], errors="coerce")
    df["annee"] = df["date_mutation"].dt.year
    df["mois"] = df["date_mutation"].dt.month
    df["trimestre"] = df["date_mutation"].dt.quarter
    df["mois_sin"] = np.sin(2 * np.pi * df["mois"] / 12)
    df["mois_cos"] = np.cos(2 * np.pi * df["mois"] / 12)

df_model = df[df["prix_m2"] > 0].copy()

 
# Filtrage de la cible — version 2
 
# La borne basse reste fondée sur le quantile 1 %, afin d'écarter
# les valeurs anormalement faibles.
# La borne haute n'est plus le quantile 99 % : elle devient une
# borne métier fixe à 45 000 €/m² pour conserver les marchés très chers
# tout en excluant les valeurs manifestement extrêmes.

PRIX_M2_MAX_METIER = 45_000

q_low = df_model["prix_m2"].quantile(0.01)
q_high = PRIX_M2_MAX_METIER

df_model = df_model[df_model["prix_m2"].between(q_low, q_high)].copy()
df_model["log_prix_m2"] = np.log(df_model["prix_m2"])

print("Lignes modèle :", f"{len(df_model):,}")
print("Borne basse prix m² conservée :", round(q_low, 0), "€/m²")
print("Borne haute prix m² conservée :", f"{PRIX_M2_MAX_METIER:,.0f}".replace(",", " "), "€/m²")
display(df_model[["prix_m2", "log_prix_m2"]].describe().T)

plt.figure(figsize=(8, 4))
plt.hist(df_model["prix_m2"], bins=60)
plt.title("Distribution du prix au m² filtré")
plt.xlabel("Prix au m²")
plt.ylabel("Volume")
save_fig("01_distribution_prix_m2.png")

plt.figure(figsize=(8, 4))
plt.hist(df_model["log_prix_m2"], bins=60)
plt.title("Distribution de log(prix au m²)")
plt.xlabel("log(prix au m²)")
plt.ylabel("Volume")
save_fig("02_distribution_log_prix_m2.png")
```

```python
# Chargement des blocs de variables définis dans le notebook 11.
# Le notebook 12 consomme directement le contrat produit par le notebook 11.
# Aucun fallback historique n'est conservé afin de garantir :
# - la reproductibilité de la chaîne,
# - l'alignement exact entre sélection amont et modélisation aval,
# - la détection explicite des ruptures de pipeline.

with open(INPUT_FEATURE_BLOCKS_PATH, "r", encoding="utf-8") as f:
    feature_blocks = json.load(f)

# Conservation de la source utilisée pour audit / traçabilité.
feature_blocks_source = str(INPUT_FEATURE_BLOCKS_PATH)

print("Blocs chargés depuis le notebook 11 :", INPUT_FEATURE_BLOCKS_PATH)

# Certains blocs peuvent contenir des variables absentes de df_model
# (suppression amont, changement de feature engineering, filtre métier, etc.).
#
# On conserve uniquement les variables effectivement présentes afin :
# - d'éviter les erreurs de pipeline,
# - de documenter les écarts entre sélection théorique et données disponibles,
# - de conserver un audit explicite des pertes de variables.

available_blocks = {
    block: [c for c in cols if c in df_model.columns]
    for block, cols in feature_blocks.items()
}

# Construction du reporting de cohérence des blocs.
#
# Ce tableau permet d'analyser :
# - le nombre de variables prévues par notebook 11,
# - le nombre réellement disponibles dans notebook 12,
# - les variables perdues entre les étapes.
#
# Les blocs vides sont volontairement conservés dans le reporting
# afin de maintenir une vision exhaustive du contrat amont.

block_summary = pd.DataFrame([
    {
        "bloc": block,

        # Nombre de variables définies dans le notebook 11.
        "nb_variables_notebook_11": len(cols),

        # Nombre de variables réellement présentes dans df_model.
        "nb_variables_disponibles_notebook_12": len(
            available_blocks.get(block, [])
        ),

        # Liste des variables effectivement exploitables.
        "variables_disponibles": ", ".join(
            available_blocks.get(block, [])
        ),

        # Variables attendues mais absentes.
        "variables_absentes": ", ".join(
            [c for c in cols if c not in df_model.columns]
        ),
    }
    for block, cols in feature_blocks.items()
])

# Ajout de la source des blocs pour audit / reproductibilité.
block_summary["source_blocs"] = feature_blocks_source

display(block_summary)

# Export du reporting pour documentation externe.
block_summary.to_csv(
    TABLE_DIR / "block_summary_depuis_notebook_11.csv",
    index=False
)

# Comptage du nombre total de variables distinctes disponibles
# après nettoyage des blocs.
print(
    "Total features disponibles via blocs notebook 11 :",
    len(set(sum(available_blocks.values(), [])))
)

# Vérification explicite du bloc B6 (taux de crédit),
# souvent critique dans les analyses immobilières.
if "B6_taux_credit" in available_blocks:
    print(
        "B6 taux crédit disponible :",
        len(available_blocks["B6_taux_credit"]),
        "variable(s)"
    )
else:
    print("B6_taux_credit absent du contrat notebook 11.")

# Fusion de toutes les variables disponibles dans une liste unique,
# triée pour garantir stabilité et reproductibilité.
all_candidate_features = sorted(
    set(sum(available_blocks.values(), []))
)

# Construction d'un audit feature-level permettant :
# - d'analyser les types de données,
# - mesurer la complétude,
# - détecter les variables quasi constantes,
# - préparer les étapes de preprocessing.

audit_rows = []

for col in all_candidate_features:

    # Série courante analysée.
    s = df_model[col]

    audit_rows.append({

        # Nom de la variable.
        "variable": col,

        # Bloc d'appartenance.
        "bloc": next(
            (b for b, cols in available_blocks.items() if col in cols),
            "autre"
        ),

        # Type pandas brut.
        "dtype": str(s.dtype),

        # Taux de valeurs manquantes.
        "pct_manquant": s.isna().mean(),

        # Nombre de modalités / valeurs distinctes.
        "nb_valeurs_distinctes": s.nunique(dropna=True),

        # Typologie simplifiée pour pilotage preprocessing.
        "type_feature": (
            "numérique"
            if pd.api.types.is_numeric_dtype(s)
            else "catégorielle"
        ),
    })

# Construction du DataFrame d'audit final.
audit_features = (
    pd.DataFrame(audit_rows)
    .sort_values(["bloc", "variable"])
)

display(audit_features)

# Export du diagnostic des features.
audit_features.to_csv(
    TABLE_DIR / "audit_features_modelisation.csv",
    index=False
)

# Sélection des 25 variables avec le plus fort taux de valeurs manquantes.
plot_missing = (
    audit_features
    .sort_values("pct_manquant", ascending=False)
    .head(25)
    .sort_values("pct_manquant")
)

# Visualisation horizontale pour faciliter la lecture des noms longs.
plt.figure(figsize=(8, 6))

plt.barh(
    plot_missing["variable"],
    plot_missing["pct_manquant"]
)

plt.title("Variables candidates les plus incomplètes")
plt.xlabel("Part de valeurs manquantes")

# Sauvegarde de la figure dans le dossier projet.
save_fig("03_completude_features.png")
```

```python
 
# Sélection initiale des features — entrées issues du notebook 11
 
# Le notebook 11 produit les blocs/scénarios de features admissibles.
# Le notebook 12 ne refait pas cette sélection amont.
# Il vérifie uniquement que les variables sont présentes, non constantes,
# puis audite leur signal statistique pour la modélisation.

selected_features = []
selection_rows = []

for _, row in audit_features.iterrows():
    col = row["variable"]

    present = col in df_model.columns
    too_constant = row["nb_valeurs_distinctes"] <= 1
    keep = present and not too_constant

    if keep:
        selected_features.append(col)

    selection_rows.append({
        "variable": col,
        "bloc": row["bloc"],
        "pct_manquant": row["pct_manquant"],
        "nb_valeurs_distinctes": row["nb_valeurs_distinctes"],
        "type_feature": row["type_feature"],
        "retenue_selection_initiale_notebook_12": keep,
        "motif_notebook_12": (
            "retenue_depuis_notebook_11"
            if keep
            else "constante_ou_absente"
        ),
        "source_selection": feature_blocks_source,
    })

# Déduplication en préservant l'ordre des blocs du notebook 11.
selected_features = list(dict.fromkeys(selected_features))
selection_report = pd.DataFrame(selection_rows)

display(
    selection_report.sort_values(
        ["retenue_selection_initiale_notebook_12", "bloc"],
        ascending=[False, True]
    )
)

selection_report.to_csv(
    TABLE_DIR / "selection_initiale_features_depuis_notebook_11.csv",
    index=False
)

print("Features candidates issues du notebook 11 :", len(all_candidate_features))
print("Features retenues pour modélisation notebook 12 :", len(selected_features))


 
# Audit du signal numérique
 
# Cet audit sert uniquement à documenter le potentiel explicatif des variables.
# Il ne remplace pas l'évaluation des modèles.
# L'information mutuelle est calculée sur échantillon pour éviter
# un temps de calcul excessif sur le dataset DVF complet.

num_selected = [
    c for c in selected_features
    if c in df_model.columns and pd.api.types.is_numeric_dtype(df_model[c])
]

signal_rows = []

for col in num_selected:
    s = pd.to_numeric(df_model[col], errors="coerce")

    if s.nunique(dropna=True) > 1:
        signal_rows.append({
            "variable": col,
            "correlation_log_prix": s.corr(df_model["log_prix_m2"]),
            "variance": s.var(),
            "pct_manquant": s.isna().mean(),
        })

signal_audit = pd.DataFrame(signal_rows)


 
# Information mutuelle — version optimisée
 

MI_SAMPLE_SIZE = 50_000

if not signal_audit.empty:
    mi_cols = signal_audit["variable"].tolist()

    df_mi_source = df_model[mi_cols + ["log_prix_m2"]].replace(
        [np.inf, -np.inf],
        np.nan
    )

    df_mi_source = df_mi_source.dropna(subset=["log_prix_m2"])

    if len(df_mi_source) > MI_SAMPLE_SIZE:
        df_mi = df_mi_source.sample(
            n=MI_SAMPLE_SIZE,
            random_state=42
        )
    else:
        df_mi = df_mi_source.copy()

    X_mi = df_mi[mi_cols].copy()
    X_mi = X_mi.fillna(X_mi.median(numeric_only=True))

    y_mi = df_mi["log_prix_m2"]

    mi_values = mutual_info_regression(
        X_mi,
        y_mi,
        random_state=42,
        n_neighbors=3
    )

    mi_audit = pd.DataFrame({
        "variable": mi_cols,
        "information_mutuelle": mi_values
    })

    signal_audit = signal_audit.merge(
        mi_audit,
        on="variable",
        how="left"
    )

    signal_audit["abs_corr"] = signal_audit["correlation_log_prix"].abs()
    signal_audit = signal_audit.sort_values(
        "information_mutuelle",
        ascending=False
    )

display(signal_audit)


 
# Visualisation et export
 

if not signal_audit.empty:
    plot_mi = signal_audit.sort_values("information_mutuelle").tail(20)

    plt.figure(figsize=(8, 5))
    plt.barh(plot_mi["variable"], plot_mi["information_mutuelle"])
    plt.title("Information mutuelle des variables numériques — échantillon")
    plt.xlabel("Score")
    save_fig("04_information_mutuelle_features.png")

signal_audit.to_csv(
    TABLE_DIR / "signal_features_numeriques.csv",
    index=False
)

print("Variables numériques auditées :", len(num_selected))
print("Taille dataset complet :", len(df_model))
print("Taille utilisée pour information mutuelle :", min(len(df_model), MI_SAMPLE_SIZE))
```

```python
# Validation méthodologique complémentaire des features.
# Les variables numériques sont discrétisées en quantiles pour tester si elles séparent les distributions de prix.
from scipy.stats import kruskal, f_oneway

feature_tests = []

for col in selected_features:
    if col not in df_model.columns:
        continue

    s = df_model[col]
    y = df_model["log_prix_m2"]

    if pd.api.types.is_numeric_dtype(s):
        valid = pd.DataFrame({"x": s, "y": y}).dropna()

        if valid["x"].nunique() >= 4 and len(valid) >= 100:
            try:
                bins = pd.qcut(valid["x"], q=4, duplicates="drop")
                groups = [
                    valid.loc[bins == g, "y"].values
                    for g in bins.dropna().unique()
                    if len(valid.loc[bins == g, "y"]) >= 30
                ]

                if len(groups) >= 2:
                    stat_kw, p_kw = kruskal(*groups)
                    stat_anova, p_anova = f_oneway(*groups)

                    feature_tests.append({
                        "variable": col,
                        "type_test": "numérique_discrétisée",
                        "nb_groupes": len(groups),
                        "kruskal_stat": stat_kw,
                        "kruskal_p_value": p_kw,
                        "anova_stat": stat_anova,
                        "anova_p_value": p_anova,
                    })
            except Exception:
                pass

    else:
        valid = pd.DataFrame({"x": s.astype("string"), "y": y}).dropna()
        counts = valid["x"].value_counts()
        kept_levels = counts[counts >= 30].index
        valid = valid[valid["x"].isin(kept_levels)]

        groups = [
            sub["y"].values
            for _, sub in valid.groupby("x")
            if len(sub) >= 30
        ]

        if len(groups) >= 2:
            stat_kw, p_kw = kruskal(*groups)
            stat_anova, p_anova = f_oneway(*groups)

            feature_tests.append({
                "variable": col,
                "type_test": "catégorielle",
                "nb_groupes": len(groups),
                "kruskal_stat": stat_kw,
                "kruskal_p_value": p_kw,
                "anova_stat": stat_anova,
                "anova_p_value": p_anova,
            })

feature_tests_df = pd.DataFrame(feature_tests)

if not feature_tests_df.empty:
    feature_tests_df = feature_tests_df.sort_values("kruskal_p_value")
    display(feature_tests_df.head(30))
    feature_tests_df.to_csv(TABLE_DIR / "validation_statistique_features.csv", index=False)
else:
    print("Aucun test statistique complémentaire exploitable.")
```

```python
# Fonction utilitaire : conserve uniquement les variables réellement présentes
# dans le DataFrame fourni.
#
# Objectif :
# - éviter les erreurs si une variable attendue est absente,
# - supprimer les doublons,
# - garder l'ordre initial des variables.
def keep_existing(features, df):
    """Retourne uniquement les variables présentes dans le DataFrame fourni, sans doublons."""
    return list(dict.fromkeys([f for f in features if f in df.columns]))


# Fonction utilitaire : récupère un bloc de variables produit par le notebook 11.
#
# Si le bloc demandé n'existe pas, la fonction retourne une liste vide.
# Cela évite une erreur bloquante et rend le pipeline plus robuste.
def block(name):
    """Accès robuste aux blocs exportés par le notebook 11."""
    return available_blocks.get(name, [])


# Fonction utilitaire : fusionne plusieurs listes de variables.
#
# Objectif :
# - concaténer plusieurs blocs,
# - supprimer les doublons,
# - ignorer les valeurs vides éventuelles,
# - conserver l'ordre d'apparition.
def uniq_features(*lists):
    """Concatène plusieurs listes de features en supprimant les doublons et les valeurs vides."""
    out = []
    for lst in lists:
        for col in lst:
            if col and col not in out:
                out.append(col)
    return out


# df_ref sert de référence unique pour vérifier l'existence des variables.
# Ici, on utilise df_model, c'est-à-dire le jeu de données préparé pour la modélisation.
df_ref = df_model


# Récupération sécurisée des blocs de variables.
#
# Chaque bloc provient du notebook 11, puis est filtré pour ne conserver
# que les colonnes réellement disponibles dans df_model.
bloc_bien = keep_existing(block("B1_bien"), df_ref)
bloc_temps = keep_existing(block("B2_temps"), df_ref)
bloc_territoire = keep_existing(block("B3_territoire"), df_ref)
bloc_socio_eco = keep_existing(block("B4_socio_eco"), df_ref)
bloc_comparables = keep_existing(block("B5_comparables"), df_ref)
bloc_taux = keep_existing(block("B6_taux_credit"), df_ref)


# Chargement des scénarios de modélisation exportés par le notebook 11.
#
# Un scénario correspond à un ensemble de variables à tester ensemble.
# Exemple :
# - scénario avec uniquement les caractéristiques du bien,
# - scénario avec bien + territoire,
# - scénario complet,
# - scénario avec ou sans variables de taux.
with open(INPUT_SCENARIOS_JSON_PATH, "r", encoding="utf-8") as f:
    scenario_features_raw = json.load(f)

# Conservation du chemin source pour assurer la traçabilité du pipeline.
scenario_source = str(INPUT_SCENARIOS_JSON_PATH)


# Nettoyage des scénarios.
#
# Pour chaque scénario, on conserve uniquement les variables présentes
# dans df_model afin d'éviter toute erreur lors de l'entraînement.
scenario_features = {
    name: keep_existing(features, df_ref)
    for name, features in scenario_features_raw.items()
}


# Sécurisation du bloc B6 : taux de crédit.
#
# Si les variables de taux sont disponibles, on garantit :
# - l'existence d'un scénario utilisant uniquement les taux,
# - l'existence de variantes "_avec_taux" pour chaque scénario existant,
# - l'existence d'un scénario complet incluant tous les blocs + les taux.
#
# Cette étape permet de comparer proprement les modèles :
# - sans variables macro-financières,
# - avec variables macro-financières.
if len(bloc_taux) > 0:

    # Création d'un scénario composé uniquement des variables de taux.
    scenario_features.setdefault(
        "S_B6_taux_only",
        keep_existing(bloc_taux, df_ref)
    )

    # Pour chaque scénario existant, création automatique d'une version enrichie
    # avec les variables de taux, sauf si cette version existe déjà.
    for scenario_name, feature_list in list(scenario_features.items()):

        # On évite de retraiter les scénarios déjà enrichis avec les taux
        # ou le scénario pur taux.
        if scenario_name.endswith("_avec_taux") or scenario_name == "S_B6_taux_only":
            continue

        scenario_features.setdefault(
            f"{scenario_name}_avec_taux",
            keep_existing(
                uniq_features(feature_list, bloc_taux),
                df_ref
            ),
        )

    # Création d'un scénario complet enrichi avec les taux.
    #
    # Il regroupe :
    # - caractéristiques du bien,
    # - variables temporelles,
    # - variables territoriales,
    # - variables socio-économiques,
    # - comparables immobiliers,
    # - taux de crédit.
    scenario_features.setdefault(
        "S_FULL_SELECTED_avec_taux",
        keep_existing(
            uniq_features(
                bloc_bien,
                bloc_temps,
                bloc_territoire,
                bloc_socio_eco,
                bloc_comparables,
                bloc_taux,
            ),
            df_ref,
        ),
    )


# Suppression des scénarios vides.
#
# Un scénario vide n'est pas exploitable pour entraîner un modèle.
# Cette étape évite donc de lancer inutilement une modélisation impossible.
scenario_features = {
    name: feats
    for name, feats in scenario_features.items()
    if len(feats) > 0
}


# Création d'un tableau de synthèse des scénarios.
#
# Ce tableau permet de vérifier :
# - le nom de chaque scénario,
# - le nombre de variables utilisées,
# - la présence ou non des variables de taux,
# - la liste complète des features,
# - les fichiers sources utilisés.
scenario_summary = pd.DataFrame([
    {
        "scenario": name,
        "nb_features": len(features),
        "integre_taux": any(f in bloc_taux for f in features),
        "features": ", ".join(features),
        "source_scenarios": scenario_source,
        "source_blocs": feature_blocks_source,
    }
    for name, features in scenario_features.items()
])

display(scenario_summary)


# Export de la synthèse des scénarios pour documentation et audit.
scenario_summary.to_csv(
    TABLE_DIR / "scenario_features_depuis_notebook_11.csv",
    index=False
)


# Messages de contrôle affichés dans le notebook.
#
# Ils permettent de vérifier rapidement :
# - la cible retenue,
# - la consommation correcte des scénarios issus du notebook 11,
# - la disponibilité des comparables,
# - la disponibilité des variables de taux,
# - la taille de chaque scénario.
print("La cible retenue est log_prix_m2 pour réduire l'asymétrie des prix.")
print("Les scénarios sont consommés depuis le notebook 11 et incluent B6 lorsque disponible.")
print("B5 comparables disponible :", len(bloc_comparables), "variable(s)")
print("B6 taux crédit disponible :", len(bloc_taux), "variable(s)")

for scenario, feats in scenario_features.items():
    print(f"- {scenario}: {len(feats)} variables")


# Construction du split train/test.
#
# Paramètre méthodologique de cette version :
# - train sur les années antérieures ;
# - test strictement sur l'année 2025.
#
# Cette variante sert à mesurer la prédictibilité temporelle du modèle,
# par opposition au split aléatoire qui mesure surtout le pouvoir explicatif
# moyen des variables sur une photographie globale du marché.
TEST_YEAR = 2025

# Récupération robuste de l'année de mutation.
#
# Selon les exports du notebook 11, l'année peut être déjà présente sous forme
# numérique, ou seulement récupérable à partir d'une colonne de date.
# On évite donc de dépendre d'un seul nom de colonne.
YEAR_COLUMN_CANDIDATES = [
    "annee",
    "annee_mutation",
    "year",
    "year_mutation",
    "annee_vente",
    "annee_transaction",
]

DATE_COLUMN_CANDIDATES = [
    "date_mutation",
    "date_vente",
    "date_transaction",
    "date",
    "date_mutation_dt",
]

# 1. Cas le plus simple : une colonne année existe déjà.
year_col = next(
    (
        c for c in YEAR_COLUMN_CANDIDATES
        if c in df_model.columns
        and pd.to_numeric(df_model[c], errors="coerce").notna().sum() > 0
    ),
    None,
)

if year_col is not None:
    df_model["annee_split"] = pd.to_numeric(df_model[year_col], errors="coerce")
    split_source = year_col

# 2. Sinon, on tente de reconstituer l'année depuis une colonne date.
else:
    date_col = next(
        (
            c for c in DATE_COLUMN_CANDIDATES
            if c in df_model.columns
            and pd.to_datetime(df_model[c], errors="coerce").notna().sum() > 0
        ),
        None,
    )

    if date_col is not None:
        df_model["annee_split"] = pd.to_datetime(
            df_model[date_col],
            errors="coerce"
        ).dt.year
        split_source = date_col

    # 3. Dernier recours : recherche prudente parmi les colonnes contenant
    #    date / annee / year, sans supposer le nom exact produit en amont.
    else:
        temporal_candidates = [
            c for c in df_model.columns
            if any(token in c.lower() for token in ["annee", "year", "date"])
        ]

        split_source = None
        for c in temporal_candidates:
            values_num = pd.to_numeric(df_model[c], errors="coerce")
            if values_num.between(2020, TEST_YEAR).sum() > 0:
                df_model["annee_split"] = values_num
                split_source = c
                break

            values_date = pd.to_datetime(df_model[c], errors="coerce")
            if values_date.notna().sum() > 0:
                candidate_year = values_date.dt.year
                if candidate_year.between(2020, TEST_YEAR).sum() > 0:
                    df_model["annee_split"] = candidate_year
                    split_source = c
                    break

        if split_source is None:
            raise ValueError(
                "Split temporel 2025 impossible : aucune colonne temporelle exploitable. "
                f"Colonnes temporelles candidates trouvées : {temporal_candidates}"
            )

# Sécurisation : l'année doit être numérique et cohérente avec le périmètre DVF.
df_model["annee_split"] = pd.to_numeric(df_model["annee_split"], errors="coerce")
df_model = df_model[df_model["annee_split"].notna()].copy()
df_model["annee_split"] = df_model["annee_split"].astype(int)

train_df = df_model[df_model["annee_split"] < TEST_YEAR].copy()
test_df = df_model[df_model["annee_split"] == TEST_YEAR].copy()

if len(train_df) < 1000 or len(test_df) < 200:
    repartition_annees = df_model["annee_split"].value_counts().sort_index()
    raise ValueError(
        f"Split temporel 2025 trop fragile : train={len(train_df):,}, "
        f"test={len(test_df):,}. Répartition disponible par année : "
        f"{repartition_annees.to_dict()}"
    )

split_type = f"temporel : train < {TEST_YEAR}, test = {TEST_YEAR} via {split_source}"

# Définition des cibles.
#
# TARGET :
# - cible transformée en log,
# - utilisée pour l'entraînement,
# - réduit l'effet des prix extrêmes.
#
# TARGET_RAW :
# - cible brute en euros/m²,
# - utile pour interpréter les performances dans l'échelle métier.
TARGET = "log_prix_m2"
TARGET_RAW = "prix_m2"


# Séparation des cibles train et test.
y_train = train_df[TARGET]
y_test = test_df[TARGET]


# Affichage du type de split et des volumes obtenus.
print("Split :", split_type)
print("Train :", f"{len(train_df):,}")
print("Test  :", f"{len(test_df):,}")
```

  <div style="color: #4b5563; font-size: 14px;">Cette cellule contient le vrai preprocessing de modélisation. Elle construit un  ColumnTransformer  avec traitement différencié des variables numériques et catégorielles, puis prépare les modèles candidats. Numérique : imputation par médiane, avec standardisation optionnelle pour les modèles linéaires. Catégoriel : imputation de la modalité manquante puis encodage one-hot. Modèles testés : baseline, linéaires régularisés, robuste, forêt aléatoire et boosting.</div>

```python
# Fonction utilitaire pour créer un OneHotEncoder compatible
# avec plusieurs versions de scikit-learn.
#
# Certaines versions utilisent :
# - sparse_output=False
# d'autres :
# - sparse=False
#
# Cette fonction garantit donc la compatibilité du notebook
# quel que soit l'environnement d'exécution.
def make_onehot():

    try:
        # Version récente de scikit-learn.
        return OneHotEncoder(
            handle_unknown="ignore",
            min_frequency=0.01,
            sparse_output=False
        )

    except TypeError:
        # Fallback pour anciennes versions.
        return OneHotEncoder(
            handle_unknown="ignore",
            min_frequency=0.01,
            sparse=False
        )


# Fonction de construction du pipeline de preprocessing.
#
# Objectifs :
# - identifier automatiquement les variables numériques et catégorielles,
# - appliquer les traitements adaptés à chaque type,
# - préparer les données pour les modèles machine learning.
def build_preprocessor(X, scale_numeric=False):

    # Détection automatique des colonnes numériques.
    num_cols = X.select_dtypes(include=[np.number]).columns.tolist()

    # Détection automatique des colonnes catégorielles.
    cat_cols = X.select_dtypes(exclude=[np.number]).columns.tolist()


    # Pipeline des variables numériques.
    #
    # Étape 1 :
    # - remplacement des valeurs manquantes par la médiane,
    # - robuste aux valeurs extrêmes.
    num_steps = [
        ("imputer", SimpleImputer(strategy="median"))
    ]

    # Optionnel :
    # ajout d'une standardisation des variables numériques.
    #
    # Utile pour :
    # - régression linéaire,
    # - Ridge,
    # - Lasso,
    # - Huber.
    #
    # Peu utile pour :
    # - RandomForest,
    # - Gradient Boosting.
    if scale_numeric:
        num_steps.append(
            ("scaler", StandardScaler())
        )


    # Pipeline des variables catégorielles.
    cat_steps = [

        # Imputation des valeurs manquantes
        # avec la modalité la plus fréquente.
        ("imputer", SimpleImputer(strategy="most_frequent")),

        # Encodage One-Hot :
        # transformation des catégories en variables binaires.
        ("onehot", make_onehot()),
    ]


    # Construction du préprocesseur global.
    #
    # ColumnTransformer applique automatiquement :
    # - le pipeline numérique aux colonnes numériques,
    # - le pipeline catégoriel aux colonnes catégorielles.
    preprocessor = ColumnTransformer(
        transformers=[

            # Pipeline numérique.
            ("num", Pipeline(num_steps), num_cols),

            # Pipeline catégoriel.
            ("cat", Pipeline(cat_steps), cat_cols),
        ],

        # Les colonnes non listées sont supprimées.
        remainder="drop",

        # Force une sortie dense plutôt que sparse matrix.
        sparse_threshold=0.0,
    )

    # Retour du pipeline + listes de colonnes.
    return preprocessor, num_cols, cat_cols


# Fonction d'évaluation des prédictions.
#
# Le modèle prédit ici :
# - log(prix_m2)
#
# Cette fonction :
# - sécurise les prédictions,
# - reconvertit les valeurs en euros/m²,
# - calcule plusieurs métriques de performance.
def evaluate_predictions(y_true_log, y_pred_log):

    # Définition de bornes de clipping.
    #
    # Objectif :
    # empêcher certaines prédictions aberrantes
    # produites par les modèles linéaires.
    lower = float(y_train.quantile(0.001) - 0.25)
    upper = float(y_train.quantile(0.999) + 0.25)

    # Clipping des prédictions.
    y_pred_log = np.clip(y_pred_log, lower, upper)


    # Retour à l'échelle métier réelle.
    #
    # Les modèles travaillent sur log(prix_m2),
    # mais l'évaluation métier se fait en euros/m².
    y_true = np.exp(y_true_log)
    y_pred = np.exp(y_pred_log)


    # Calcul des métriques de performance.
    return {

        # Erreur absolue moyenne en euros/m².
        "MAE": float(mean_absolute_error(y_true, y_pred)),

        # Racine de l'erreur quadratique moyenne.
        "RMSE": float(
            mean_squared_error(
                y_true,
                y_pred,
                squared=False
            )
        ),

        # Coefficient de détermination sur échelle réelle.
        "R2": float(r2_score(y_true, y_pred)),

        # Coefficient de détermination sur échelle log.
        "R2_log": float(r2_score(y_true_log, y_pred_log)),

        # Valeurs réelles en euros/m².
        "y_true_euros": y_true,

        # Prédictions reconverties en euros/m².
        "y_pred_euros": y_pred,

        # Prédictions conservées sur échelle log.
        "y_pred_log": y_pred_log,
    }


print("Préprocessing prêt")


# Structures de stockage des résultats.
#
# results :
# - tableau récapitulatif des performances.
#
# fitted_models :
# - stockage des modèles entraînés.
#
# predictions :
# - stockage des prédictions produites.
results = []
fitted_models = {}
predictions = {}


# Définition des modèles à tester.
#
# Chaque entrée contient :
# - le nom du modèle,
# - l'objet modèle scikit-learn,
# - la famille du modèle,
# - si la standardisation est nécessaire,
# - une limite éventuelle de taille d'échantillon.
model_specs = [

    # Baseline naïve :
    # prédit toujours la médiane.
    ("Dummy_median",
     DummyRegressor(strategy="median"),
     "baseline",
     False,
     None),

    # Régression linéaire classique.
    ("LinearRegression",
     LinearRegression(),
     "linear",
     True,
     None),

    # Ridge :
    # régression linéaire régularisée L2.
    ("Ridge_alpha_1",
     Ridge(alpha=1.0),
     "linear",
     True,
     None),

    # Lasso :
    # régression linéaire régularisée L1.
    # Peut effectuer une sélection automatique de variables.
    ("Lasso_alpha_0_001",
     Lasso(alpha=0.001, max_iter=5000),
     "linear",
     True,
     None),

    # Huber :
    # régression robuste aux valeurs extrêmes.
    ("HuberRegressor",
     HuberRegressor(max_iter=300),
     "robust",
     True,
     120000),

    # Random Forest :
    # ensemble d'arbres de décision aléatoires.
    ("RandomForest",
     RandomForestRegressor(
         n_estimators=80,
         min_samples_leaf=2,
         max_features="sqrt",
         random_state=42,
         n_jobs=-1,
     ),
     "tree",
     False,
     80000),

    # Gradient Boosting optimisé histogramme.
    #
    # Très performant sur données tabulaires volumineuses.
    ("HistGradientBoosting",
     HistGradientBoostingRegressor(
         random_state=42,
         max_iter=150,
         learning_rate=0.05,
     ),
     "boosting",
     False,
     120000),
]
```

```python
# BENCHMARK INITIAL DES SCÉNARIOS
from sklearn.base import clone
from sklearn.pipeline import Pipeline
import time
import gc

results = []
fitted_models = {}
predictions = {}

MAX_SCENARIO_FEATURES = None  # laisser None sauf debug
SAFE_TRAIN_CAP_DEFAULT = 500_000


def clean_feature_frame(X: pd.DataFrame) -> pd.DataFrame:
    """
    Nettoyage défensif avant pipeline sklearn.
    Objectif : éviter les erreurs pandas.NA / object dtype / inf.
    """
    X = X.copy()

    # Remplacement global des valeurs problématiques
    X = X.replace([np.inf, -np.inf], np.nan)
    X = X.replace({pd.NA: np.nan, None: np.nan})

    # Colonnes booléennes -> int
    bool_cols = X.select_dtypes(include=["bool", "boolean"]).columns.tolist()
    for c in bool_cols:
        X[c] = X[c].astype("float64")

    # Colonnes catégorielles / object -> string propre
    cat_like_cols = X.select_dtypes(include=["object", "category", "string"]).columns.tolist()
    for c in cat_like_cols:
        X[c] = X[c].astype("string").fillna("__MISSING__").astype(str)

    # Colonnes numériques : coercition stricte
    for c in X.columns:
        if c not in cat_like_cols:
            X[c] = pd.to_numeric(X[c], errors="coerce")

    return X


def safe_feature_list(features, df):
    """
    Garde uniquement les colonnes présentes, non dupliquées et non constantes.
    """
    cols = []
    for c in features:
        if c in df.columns and c not in cols:
            nunique = df[c].nunique(dropna=True)
            if nunique > 1:
                cols.append(c)
    return cols


def run_model(
    scenario,
    features,
    model_name,
    estimator,
    family,
    scale_numeric,
    train_cap=None,
):
    features = safe_feature_list(features, train_df)

    if MAX_SCENARIO_FEATURES is not None:
        features = features[:MAX_SCENARIO_FEATURES]

    if len(features) == 0:
        print(f"[SKIP] {scenario} | {model_name} : aucune feature exploitable.")
        return

    X_train = clean_feature_frame(train_df[features])
    X_test = clean_feature_frame(test_df[features])
    ytr = y_train.copy()

    # Sécurisation index y / X
    ytr = ytr.loc[X_train.index]

    effective_train_cap = train_cap
    if effective_train_cap is None and family in ["linear", "tree_sklearn"]:
        effective_train_cap = SAFE_TRAIN_CAP_DEFAULT

    if effective_train_cap is not None and len(X_train) > effective_train_cap:
        sample_idx = X_train.sample(effective_train_cap, random_state=42).index
        X_train = X_train.loc[sample_idx]
        ytr = ytr.loc[sample_idx]

    preprocessor, num_cols, cat_cols = build_preprocessor(
        X_train,
        scale_numeric=scale_numeric,
    )

    pipe = Pipeline([
        ("preprocessor", preprocessor),
        ("model", clone(estimator)),
    ])

    print(
        f"\n[RUN] {scenario} | {model_name} | "
        f"{len(features)} features | "
        f"{len(num_cols)} num | {len(cat_cols)} cat | "
        f"{len(X_train):,} lignes train"
    )

    start = time.time()

    try:
        pipe.fit(X_train, ytr)
    except Exception as e:
        print(f"[ERREUR] {scenario} | {model_name} : {type(e).__name__} - {e}")

        results.append({
            "scenario": scenario,
            "model": model_name,
            "family": family,
            "n_features_raw": len(features),
            "n_numeric": len(num_cols),
            "n_categorical": len(cat_cols),
            "n_train_used": len(X_train),
            "fit_time_sec": np.nan,
            "MAE_train": np.nan,
            "RMSE_train": np.nan,
            "R2_train": np.nan,
            "MAE_test": np.nan,
            "RMSE_test": np.nan,
            "R2_test": np.nan,
            "R2_log_test": np.nan,
            "ecart_R2_train_test": np.nan,
            "status": "error",
            "error_message": str(e),
        })
        return

    fit_time = time.time() - start

    pred_train_log = pipe.predict(X_train)
    pred_test_log = pipe.predict(X_test)

    train_eval = evaluate_predictions(ytr, pred_train_log)
    test_eval = evaluate_predictions(y_test, pred_test_log)

    row = {
        "scenario": scenario,
        "model": model_name,
        "family": family,
        "n_features_raw": len(features),
        "n_numeric": len(num_cols),
        "n_categorical": len(cat_cols),
        "n_train_used": len(X_train),
        "fit_time_sec": fit_time,
        "MAE_train": train_eval["MAE"],
        "RMSE_train": train_eval["RMSE"],
        "R2_train": train_eval["R2"],
        "MAE_test": test_eval["MAE"],
        "RMSE_test": test_eval["RMSE"],
        "R2_test": test_eval["R2"],
        "R2_log_test": test_eval["R2_log"],
        "ecart_R2_train_test": train_eval["R2"] - test_eval["R2"],
        "status": "ok",
        "error_message": "",
    }

    results.append(row)
    fitted_models[(scenario, model_name)] = pipe

    predictions[(scenario, model_name)] = pd.DataFrame({
        "y_true": test_eval["y_true_euros"],
        "y_pred": test_eval["y_pred_euros"],
        "y_true_log": y_test,
        "y_pred_log": test_eval["y_pred_log"],
        "residu": test_eval["y_true_euros"] - test_eval["y_pred_euros"],
    }, index=test_df.index)

    print(
        f"[OK] {scenario} | {model_name} | "
        f"RMSE test={test_eval['RMSE']:,.0f} | "
        f"MAE test={test_eval['MAE']:,.0f} | "
        f"R² test={test_eval['R2']:.4f} | "
        f"{fit_time/60:.1f} min"
    )

    gc.collect()


 
# EXÉCUTION BENCHMARK
 

for scenario, features in scenario_features.items():
    for model_name, estimator, family, scale_numeric, train_cap in model_specs:
        run_model(
            scenario=scenario,
            features=features,
            model_name=model_name,
            estimator=estimator,
            family=family,
            scale_numeric=scale_numeric,
            train_cap=train_cap,
        )

benchmark_df = pd.DataFrame(results)

if benchmark_df.empty:
    raise RuntimeError("Aucun modèle n'a été exécuté.")

benchmark_df_ok = (
    benchmark_df[benchmark_df["status"].eq("ok")]
    .sort_values(["RMSE_test", "MAE_test"])
    .reset_index(drop=True)
)

benchmark_df.to_csv(TABLE_DIR / "benchmark_initial_modeles_complet.csv", index=False)
benchmark_df_ok.to_csv(TABLE_DIR / "benchmark_initial_modeles.csv", index=False)

display(benchmark_df_ok[[
    "scenario", "model", "MAE_test", "RMSE_test", "R2_test",
    "R2_train", "ecart_R2_train_test", "fit_time_sec"
]].head(30))


 
# MEILLEUR MODÈLE PAR SCÉNARIO
best_initial = benchmark_df_ok.iloc[0]

best_by_scenario = (
    benchmark_df_ok.sort_values(["scenario", "RMSE_test"])
    .groupby("scenario")
    .head(1)
    .reset_index(drop=True)
)

display(best_by_scenario[[
    "scenario", "model", "MAE_test", "RMSE_test",
    "R2_test", "ecart_R2_train_test"
]])

best_by_scenario.to_csv(TABLE_DIR / "best_model_by_scenario.csv", index=False)


 
# VISUALISATIONS
plot_bench = best_by_scenario.sort_values("RMSE_test", ascending=True)

plt.figure(figsize=(10, 5))
plt.bar(plot_bench["scenario"], plot_bench["RMSE_test"])
plt.title("Meilleur modèle par scénario - RMSE test")
plt.ylabel("RMSE")
plt.xticks(rotation=30, ha="right")
save_fig("05_rmse_best_by_scenario.png")

plt.figure(figsize=(10, 5))
plt.bar(plot_bench["scenario"], plot_bench["R2_test"])
plt.title("Meilleur modèle par scénario - R² test")
plt.ylabel("R² test")
plt.xticks(rotation=30, ha="right")
save_fig("06_r2_best_by_scenario.png")

plt.figure(figsize=(10, 5))
plt.bar(plot_bench["scenario"], plot_bench["MAE_test"])
plt.title("Apport des blocs de features - MAE test")
plt.ylabel("MAE test")
plt.xticks(rotation=30, ha="right")
save_fig("06ter_apport_blocs_features_mae.png")


 
# SÉLECTION DES SCÉNARIOS POUR TUNING
top_scenarios = best_by_scenario.sort_values("RMSE_test").head(2)["scenario"].tolist()

tuning_results = []
tuned_models = {}

print("\nScénarios retenus pour tuning :", top_scenarios)
print("Meilleur modèle initial :", dict(best_initial))

```

```python
# Fonction principale de recherche d'hyperparamètres.
#
# Objectif :
# - entraîner un modèle sur un scénario donné,
# - tester plusieurs configurations d'hyperparamètres,
# - sélectionner automatiquement la meilleure version du modèle,
# - évaluer les performances sur le jeu de test.
def run_search(
    scenario,
    model_label,
    base_estimator,
    param_grid,
    search_type="grid",
    n_iter=8,
    train_cap=120_000,
    cv=3,
    n_jobs=1,
):

    # Récupération sécurisée des variables du scénario.
    #
    # Cette fonction garantit que :
    # - les colonnes existent réellement,
    # - aucune variable invalide n'est utilisée.
    features = safe_feature_list(
        scenario_features[scenario],
        train_df
    )


    # Construction du DataFrame d'entraînement.
    #
    # clean_feature_frame applique les nettoyages nécessaires :
    # - gestion des types,
    # - suppression d'anomalies,
    # - harmonisation éventuelle des colonnes.
    X_train_full = clean_feature_frame(train_df[features])

    # Alignement de la cible avec les lignes conservées.
    y_train_full = y_train.loc[X_train_full.index].copy()


    # Limitation éventuelle de la taille du train.
    #
    # Objectif :
    # - réduire les temps de calcul,
    # - limiter la consommation mémoire,
    # - accélérer le tuning.
    #
    # Un échantillonnage aléatoire reproductible est utilisé.
    if train_cap is not None and len(X_train_full) > train_cap:

        sample_idx = X_train_full.sample(
            train_cap,
            random_state=42
        ).index

        X_train_full = X_train_full.loc[sample_idx]
        y_train_full = y_train_full.loc[sample_idx]


    # Détermination automatique du besoin de standardisation.
    #
    # Les modèles linéaires régularisés
    # (Ridge / Lasso)
    # nécessitent des variables mises à l'échelle.
    scale_numeric = model_label.startswith(("Ridge", "Lasso"))


    # Construction du pipeline de preprocessing.
    preprocessor, _, _ = build_preprocessor(
        X_train_full,
        scale_numeric=scale_numeric,
    )


    # Création du pipeline complet :
    # preprocessing + modèle.
    #
    # Cela garantit que :
    # - toutes les transformations sont reproduites identiquement,
    # - aucune fuite de données ne survient,
    # - le pipeline est compatible cross-validation.
    pipe = Pipeline([
        ("preprocessor", preprocessor),
        ("model", clone(base_estimator)),
    ])


    # Choix de la stratégie de recherche.
    #
    # Deux possibilités :
    #
    # 1. GridSearch :
    #    teste exhaustivement toutes les combinaisons.
    #
    # 2. RandomizedSearch :
    #    teste un sous-ensemble aléatoire plus rapide.
    if search_type == "random":

        search = RandomizedSearchCV(

            # Pipeline complet à optimiser.
            pipe,

            # Distribution des hyperparamètres.
            param_distributions=param_grid,

            # Nombre de configurations testées.
            n_iter=n_iter,

            # Métrique d'optimisation :
            # RMSE en espace log.
            scoring="neg_root_mean_squared_error",

            # Nombre de folds de validation croisée.
            cv=cv,

            # Reproductibilité.
            random_state=42,

            # Nombre de processus parallèles.
            n_jobs=n_jobs,

            # Affichage progression.
            verbose=1,
        )

    else:

        search = GridSearchCV(

            pipe,

            # Grille exhaustive de paramètres.
            param_grid=param_grid,

            scoring="neg_root_mean_squared_error",

            cv=cv,

            n_jobs=n_jobs,

            verbose=1,
        )


    # Message de suivi affiché dans le notebook.
    print(
        f"\n[TUNING] {scenario} | {model_label} | "
        f"{len(X_train_full):,} lignes | "
        f"{len(features)} features"
    )


    # Mesure du temps d'entraînement.
    start = time.time()

    # Lancement du tuning.
    search.fit(X_train_full, y_train_full)

    duration = time.time() - start


    # Préparation du jeu de test.
    X_test = clean_feature_frame(test_df[features])


    # Prédictions du meilleur modèle trouvé.
    pred_log = search.best_estimator_.predict(X_test)


    # Évaluation complète des performances.
    test_eval = evaluate_predictions(y_test, pred_log)


    # Construction de la ligne de reporting.
    row = {

        # Scénario testé.
        "scenario": scenario,

        # Nom du modèle.
        "model": model_label,

        # Meilleurs hyperparamètres trouvés.
        "best_params": search.best_params_,

        # Score cross-validation.
        "cv_rmse_log": -search.best_score_,

        # Métriques sur jeu de test.
        "MAE_test": test_eval["MAE"],
        "RMSE_test": test_eval["RMSE"],
        "R2_test": test_eval["R2"],
        "R2_log_test": test_eval["R2_log"],

        # Temps total d'entraînement.
        "fit_time_sec": duration,
    }


    # Sauvegarde des résultats du tuning.
    tuning_results.append(row)

    # Sauvegarde du meilleur modèle entraîné.
    tuned_models[(scenario, model_label)] = search.best_estimator_


    # Sauvegarde des prédictions détaillées.
    #
    # Très utile pour :
    # - analyses d'erreurs,
    # - graphiques,
    # - diagnostics,
    # - étude des résidus.
    predictions[(scenario, model_label)] = pd.DataFrame({

        # Valeurs réelles en euros/m².
        "y_true": test_eval["y_true_euros"],

        # Valeurs prédites.
        "y_pred": test_eval["y_pred_euros"],

        # Versions log.
        "y_true_log": y_test,
        "y_pred_log": test_eval["y_pred_log"],

        # Résidu = erreur de prédiction.
        "residu": (
            test_eval["y_true_euros"]
            - test_eval["y_pred_euros"]
        ),

    }, index=test_df.index)


    # Message final de synthèse.
    print(
        f"[OK] {scenario} | {model_label} | "
        f"RMSE={test_eval['RMSE']:,.0f} | "
        f"MAE={test_eval['MAE']:,.0f} | "
        f"R²={test_eval['R2']:.4f} | "
        f"{duration/60:.1f} min"
    )


# Structures de stockage des résultats du tuning.
tuning_results = []
tuned_models = {}


# Sélection des meilleurs scénarios issus du benchmark initial.
#
# On conserve les 2 meilleurs scénarios selon le RMSE.
top_scenarios = (
    best_by_scenario
    .sort_values("RMSE_test")
    .head(2)["scenario"]
    .tolist()
)


# Recherche des scénarios contenant les variables de taux.
#
# Objectif :
# - garantir qu'au moins un scénario "avec taux"
#   soit testé lors du tuning.
taux_candidates = [
    s
    for s in best_by_scenario
        .sort_values("RMSE_test")["scenario"]
        .tolist()
    if s.endswith("_avec_taux")
    or s == "S_B6_taux_only"
]


# Si aucun scénario avec taux n'est présent dans le top,
# on ajoute le meilleur scénario avec taux.
if taux_candidates and not any(
    s in top_scenarios
    for s in taux_candidates
):
    top_scenarios = list(
        dict.fromkeys(
            top_scenarios + [taux_candidates[0]]
        )
    )[:3]


print("Scénarios retenus pour tuning :", top_scenarios)


# Boucle principale de tuning.
#
# Chaque scénario est optimisé avec :
# - un modèle Ridge,
# - un modèle HistGradientBoosting.
for scenario in top_scenarios:

    # Tuning Ridge.
    #
    # Recherche exhaustive sur plusieurs niveaux de régularisation.
    run_search(
        scenario=scenario,
        model_label="Ridge_tuned",
        base_estimator=Ridge(),
        param_grid={
            "model__alpha": [0.1, 1.0, 10.0, 50.0]
        },
        search_type="grid",
        train_cap=300_000,
        cv=3,
        n_jobs=1,
    )


    # Tuning HistGradientBoosting.
    #
    # Recherche aléatoire plus efficace
    # car l'espace d'hyperparamètres est volumineux.
    run_search(
        scenario=scenario,
        model_label="HistGradientBoosting_tuned",

        base_estimator=HistGradientBoostingRegressor(
            random_state=42
        ),

        param_grid={

            # Vitesse d'apprentissage.
            "model__learning_rate": [0.03, 0.05, 0.08],

            # Nombre maximal d'itérations.
            "model__max_iter": [150, 250],

            # Complexité des arbres.
            "model__max_leaf_nodes": [31, 63],

            # Taille minimale des feuilles.
            "model__min_samples_leaf": [30, 80],

            # Régularisation L2.
            "model__l2_regularization": [0.0, 0.1, 1.0],
        },

        search_type="random",

        # Nombre de combinaisons aléatoires testées.
        n_iter=8,

        train_cap=150_000,

        cv=3,

        n_jobs=1,
    )


# Construction du DataFrame final de tuning.
tuning_df = (
    pd.DataFrame(tuning_results)
    .sort_values(["RMSE_test", "MAE_test"])
    .reset_index(drop=True)
)

display(tuning_df)


# Export des résultats du tuning.
tuning_df.to_csv(
    TABLE_DIR / "optimisation_hyperparametres.csv",
    index=False
)


# Fusion :
# - benchmark initial,
# - modèles optimisés.
#
# Cela permet de comparer directement :
# - les modèles standards,
# - les modèles tunés.
combined_results = pd.concat([

    benchmark_df.rename(columns={
        "MAE_test": "MAE",
        "RMSE_test": "RMSE",
        "R2_test": "R2",
    })[
        ["scenario", "model", "MAE", "RMSE", "R2", "fit_time_sec"]
    ],

    tuning_df.rename(columns={
        "MAE_test": "MAE",
        "RMSE_test": "RMSE",
        "R2_test": "R2",
    })[
        ["scenario", "model", "MAE", "RMSE", "R2", "fit_time_sec"]
    ],

], ignore_index=True)


# Classement global des modèles.
combined_results = (
    combined_results
    .sort_values(["RMSE", "MAE"])
    .reset_index(drop=True)
)

display(combined_results.head(20))


# Export du benchmark consolidé.
combined_results.to_csv(
    TABLE_DIR / "benchmark_modeles_initial_et_tuning.csv",
    index=False
)
```

```python
# Création d'un DataFrame d'analyse des erreurs.
# On repart du jeu de test, puis on ajoute les prédictions du modèle final.
error_analysis = test_df.copy()

error_analysis = error_analysis.join(
    final_pred[["y_true", "y_pred", "residu", "abs_error"]],
    how="inner",
)

# Liste des variables de segmentation potentiellement disponibles.
# Ces variables permettent d'analyser les erreurs par type de bien,
# département, marché, classe de surface, etc.
segment_cols = [
    c for c in [
        "type_bien",
        "type_local",
        "code_departement",
        "typologie_marche_metier",
        "typologie_marche_finale",
        "typologie_metier",
        "phase_marche",
        "classe_surface",
    ]
    if c in error_analysis.columns
]

# Stockage des tableaux d'erreurs par segment.
segment_error_tables = []

# Pour chaque variable de segmentation,
# on calcule :
# - le nombre d'observations,
# - la MAE moyenne,
# - le prix médian réel.
for col in segment_cols:
    tab = (
        error_analysis
        .groupby(col, observed=True)
        .agg(
            n=("abs_error", "count"),
            mae=("abs_error", "mean"),
            prix_m2_median=("y_true", "median"),
        )
        .reset_index()
        .sort_values("mae", ascending=False)
    )

    # Ajout du nom de la variable de segmentation.
    tab["segment_variable"] = col

    segment_error_tables.append(tab)

    # Affichage des 10 segments les plus difficiles.
    display(tab.head(10))

# Export consolidé des erreurs par segment.
if segment_error_tables:
    segment_errors = pd.concat(segment_error_tables, ignore_index=True)

    segment_errors.to_csv(
        TABLE_DIR / "erreurs_par_segment.csv",
        index=False
    )
else:
    segment_errors = pd.DataFrame()
    print("Aucune variable de segment disponible.")
```

```python
# Récupération du préprocesseur et du modèle final depuis le pipeline.
preprocessor = final_model.named_steps["preprocessor"]
model_obj = final_model.named_steps["model"]

# Cas 1 :
# certains modèles possèdent directement un attribut feature_importances_.
# C'est le cas de nombreux modèles à base d'arbres.
if hasattr(model_obj, "feature_importances_"):

    # Noms des variables après preprocessing.
    feature_names = preprocessor.get_feature_names_out()

    # Importance fournie par le modèle.
    importance_values = model_obj.feature_importances_

else:

    # Cas 2 :
    # si le modèle ne fournit pas d'importance native,
    # on calcule une permutation importance.
    #
    # Principe :
    # - on mélange une variable,
    # - on observe la perte de performance,
    # - plus la perte est forte, plus la variable est importante.

    features_final = list(preprocessor.feature_names_in_)

    X_eval = clean_feature_frame(
    test_df.loc[:, features_final]
)

    X_eval = clean_feature_frame(test_df[features_final])

    # Échantillonnage pour limiter le temps de calcul.
    X_sample = X_eval.sample(
        n=min(4000, len(X_eval)),
        random_state=42,
    )

    y_sample = y_test.loc[X_sample.index]

    # Harmonisation des valeurs manquantes.
    X_sample = X_sample.replace({pd.NA: np.nan, None: np.nan})

    # Nettoyage des variables texte / catégorielles.
    for c in X_sample.columns:
        if pd.api.types.is_object_dtype(X_sample[c]):
            X_sample[c] = (
                X_sample[c]
                .astype("string")
                .fillna("__MISSING__")
                .astype(str)
            )

    # Calcul de l'importance par permutation.
    perm = permutation_importance(
        final_model,
        X_sample,
        y_sample,
        n_repeats=5,
        random_state=42,
        n_jobs=1,
    )

    importance_values = perm.importances_mean
    feature_names = X_sample.columns

# Tableau des 25 variables les plus importantes.
importance_df = (
    pd.DataFrame({
        "feature": feature_names,
        "importance": importance_values,
    })
    .sort_values("importance", ascending=False)
    .head(25)
)

display(importance_df)

# Graphique horizontal des importances.
plt.figure(figsize=(9, 6))
plot_imp = importance_df.iloc[::-1]

plt.barh(plot_imp["feature"], plot_imp["importance"])
plt.title("Top importances - modèle final")
plt.xlabel("Importance")
save_fig("11_importances_modele_final.png")

# Export du tableau d'importance.
importance_df.to_csv(
    TABLE_DIR / "importance_variables_modele_final.csv",
    index=False
)
```

```python
# Objectif :
# comparer le modèle final à une approche plus simple
# basée uniquement sur un indicateur de comparables immobiliers.

comparison_rows = []

# Liste de variables candidates pouvant représenter
# une estimation par comparables.
comp_candidates = [
    "prix_m2_comparables_median",
    "prix_m2_comparables_q25",
    "prix_m2_comparables_q75",
    "commune_prix_m2_med_passe",
    "commune_type_prix_m2_med_passe",
    "iris_prix_m2_med_passe",
    "geo_300m_12m_all_prix_m2_med",
    "geo_300m_24m_all_prix_m2_med",
    "geo_500m_12m_all_prix_m2_med",
    "geo_500m_24m_all_prix_m2_med",
]

# Sélection de la première variable de comparables disponible.
comp_col = next(
    (c for c in comp_candidates if c in test_df.columns),
    None
)

# Si une variable de comparables existe,
# on calcule sa MAE par rapport au vrai prix.
if comp_col is not None:
    comp_eval = test_df[["prix_m2", comp_col]].dropna().copy()

    if not comp_eval.empty:
        comparison_rows.append({
            "methode": f"comparables_{comp_col}",
            "MAE": float(
                mean_absolute_error(
                    comp_eval["prix_m2"],
                    comp_eval[comp_col]
                )
            ),
            "n": len(comp_eval),
        })

# Ajout de la performance du modèle final.
comparison_rows.append({
    "methode": f"modele_final_{final_row['model']}",
    "MAE": float(final_row["MAE"]),
    "n": len(final_pred),
})

# Tableau de comparaison.
comparison_df = pd.DataFrame(comparison_rows)

display(comparison_df)

comparison_df.to_csv(
    TABLE_DIR / "comparables_vs_modele.csv",
    index=False
)
```

```python
##  Résultats principaux — rappel du modèle final

# Cette cellule reprend les informations déjÃ  produites dans la V6 afin de fixer le point de comparaison. Le modèle final reste le modèle principal du projet ; les blocs suivants constituent des pistes d’amélioration et d’analyse.
print("Modèle final V6")
print("- Modèle   :", final_row["model"])
print("- Scénario :", final_row["scenario"])
print("- MAE      :", round(final_row["MAE"], 0), "€/m²")
print("- RMSE     :", round(final_row["RMSE"], 0), "€/m²")
print("- R²       :", round(final_row["R2"], 3))

features_best = scenario_features[final_row["scenario"]]
y_pred_base = final_pred["y_pred"]
y_true_base = final_pred["y_true"]
## 2. Tableau métriques complet

# Le R² ne suffit pas pour évaluer un modèle immobilier. La MAE donne une lecture métier en euros par mètre carré, le RMSE met en évidence les grosses erreurs, et l’erreur relative permet de replacer la performance dans l’échelle réelle du marché.
def evaluate_regression_model_complete(y_true, y_pred, model_name="Modèle"):
    # Évalue un modèle en prix réel. À utiliser après chaque variante.    y_true = pd.Series(y_true).astype(float)
    y_pred = pd.Series(y_pred, index=y_true.index).astype(float)

    mask = y_true.notna() & y_pred.notna() & (y_true > 0)
    y_true = y_true.loc[mask]
    y_pred = y_pred.loc[mask]

    errors = np.abs(y_true - y_pred)
    error_pct = errors / y_true
    residuals = y_true - y_pred

    return pd.DataFrame([{
        "modele": model_name,
        "R2": float(r2_score(y_true, y_pred)),
        "MAE_€/m2": float(mean_absolute_error(y_true, y_pred)),
        "RMSE_€/m2": float(mean_squared_error(y_true, y_pred, squared=False)),
        "Erreur_moyenne_%": float(error_pct.mean() * 100),
        "Erreur_mediane_%": float(error_pct.median() * 100),
        "Residu_moyen_€/m2": float(residuals.mean()),
        "N_test": int(len(y_true)),
    }])


def display_eval(df_eval):
    cols_round = ["R2", "MAE_€/m2", "RMSE_€/m2", "Erreur_moyenne_%", "Erreur_mediane_%", "Residu_moyen_€/m2"]
    out = df_eval.copy()
    for c in cols_round:
        if c in out.columns:
            out[c] = out[c].round(3 if c == "R2" else 1)
    display(out)


df_eval_finale = evaluate_regression_model_complete(
    y_true_base,
    y_pred_base,
    model_name=f"V6_final_{final_row['model']}_{final_row['scenario']}"
)

display_eval(df_eval_finale)
```
